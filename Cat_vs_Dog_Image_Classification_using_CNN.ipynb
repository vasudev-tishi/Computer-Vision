**Project Title:Cat vs Dog Image Classification using CNN**

Extract Dataset

```python
import zipfile

with zipfile.ZipFile('/content/archive (3).zip', 'r') as zip_ref:
    zip_ref.extractall('/content/data')

print("Dataset Extracted Successfully!")
```

Check Folder Structure

```python
import os

for root, dirs, files in os.walk('/content/data'):
    print(root)
    break
```

Import Libraries

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
```

Data Preprocessing

```python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)

test_datagen = ImageDataGenerator(rescale=1./255)
```

Load Images

```python
train_data = train_datagen.flow_from_directory(
    '/content/data/training_set/training_set',
    target_size=(64,64),
    batch_size=32,
    class_mode='binary'
)

test_data = test_datagen.flow_from_directory(
    '/content/data/test_set/test_set',
    target_size=(64,64),
    batch_size=32,
    class_mode='binary'
)
```

Build CNN Model

```python
cnn = Sequential()

cnn.add(Conv2D(32,(3,3),activation='relu',input_shape=(64,64,3)))
cnn.add(MaxPooling2D(pool_size=(2,2)))

cnn.add(Conv2D(32,(3,3),activation='relu'))
cnn.add(MaxPooling2D(pool_size=(2,2)))

cnn.add(Flatten())

cnn.add(Dense(units=128,activation='relu'))
cnn.add(Dense(units=1,activation='sigmoid'))
```

Compile Model

```python
cnn.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

Train Model

```python
history = cnn.fit(
    train_data,
    validation_data=test_data,
    epochs=5
)
```

Evaluate

```python
loss, accuracy = cnn.evaluate(test_data)

print("Accuracy:", accuracy)
```

Predict New Image

```python
from tensorflow.keras.preprocessing import image
import numpy as np

img = image.load_img(
    '/content/data/test_set/test_set/cats/cat.4001.jpg',
    target_size=(64,64)
)

img_array = image.img_to_array(img)
img_array = np.expand_dims(img_array,axis=0)
img_array = img_array/255.0

prediction = cnn.predict(img_array)

if prediction[0][0] > 0.5:
    print("Dog")
else:
    print("Cat")
```
